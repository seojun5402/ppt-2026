```python
class Counter :
    def __init__(self, initial_count=0):
        self.count = initial_count

    def increment(self, amount=1):
        self.count += amount

    def get_count(self):
        return self.count


def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    youre = Counter(5)
    print(youre.get_count())
    youre.increment()
    print(youre.get_count())
    youre.increment(4)
    print(youre.get_count())

if __name__ == "__main__":
    test_counter()
```

```python
from datetime import datetime

class Person: 

    def __init__(self, name, birth_year):
        self.__name = name
        self.__birth_year = birth_year

    def get_name(self):
        return self.__name

    def get_age(self):      
        current_year = self.get_current_year()
        return current_year - self.__birth_year

    def get_current_year(self):
        return datetime.now().year

    def __str__(self):
        return f"Name: {self.__name}, Age: {self.get_age()}"

    def test_person():
        song = Person("song", 2001)
        y = song.get_current_year()
        print(y)
        team_members = [Person("Kim", 2000), Person("Lee", 1999)]
        for member in team_members:
            print(member)

if __name__ == "__main__":
        Person.test_person()
```
