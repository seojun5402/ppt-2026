#다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1=10):
        self.attr1 = v1

    def display1(self):
        print(f"{self.attr1}")

    def display(self):
        self.display1()

class Base2:
    def __init__(self, v2=20):
        self.attr2 = v2

    def display2(self):
        print(f"{self.attr2}")

    def display(self):
        self.display2()


class C(Base1, Base2):
    def __init__(self, v1=10, v2=20, v3=30):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3


    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
        self.display()


def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2 = C(4,5,6)
    c2.print()

if __name__ == "__main__":
    test()
```

Overriding:부모의 함수(메소드) 재정의

```python
import math

class Shape:
    def __init__(self, name="도형"):
        self.name = name

    def draw(self):
            print(f"{self.name}을 그립니다.")

    def get_area(self):
            print(f"{self.name}의 면적을 계산합니다.")

class Circle(Shape):
    def __init__(self, radius=1, name="원"):
        super().__init__(name)
        self.radius = radius
    def draw(self):
            print(f"{self.name}을 그립니다.")

    def get_area(self) -> float:
            print(f"Circle {self.name}의 면적을 계산합니다.")
            return math.pi * self.radius ** 2
    
class Rectangle(Shape):
    def __init__(self, width = 1, height = 1, name = "사각형" ):
        super().__init__(name)
        self.width = width
        self.height = height

    def draw(self):
            print(f"{self.name}을 그립니다.")
    
    def get_area(self) -> float:
            print(f"Circle {self.name}의 면적을 계산합니다.")
            return self.width * self.height

   
def test():
    c = Circle(10)
    #c.draw()
    print(c.get_area())

if __name__ == "__main__":
    test()

r1 = Rectangle(width = 5, height=10)
r1.draw()
print(r1.get_area())


print("---------------------------")

print("---------------------------")
circles = [Circle(1), Circle(2), Circle(3)]
for c in circles:
    #c.draw()
    print(c.get_area())
print("---------------------------")

print("---------------------------")


rectangles = [Rectangle(1,1), Rectangle(2,2), Rectangle(3,3)]
for r in rectangles:
    #r.draw()
    print(r.get_area())
print("---------------------------")

print("---------------------------")
```
